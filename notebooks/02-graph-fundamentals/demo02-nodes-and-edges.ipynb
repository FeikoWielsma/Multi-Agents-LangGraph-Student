## a · nodes-and-edges

We pass a typed project board through two ordinary Python nodes. Each node returns only the key it changed, and
LangGraph writes that key onto the board.

```python
from typing import TypedDict
from langgraph.graph import END, START, StateGraph

class Board(TypedDict):
    brief: str
    rules: str
    code: str

def designer(board: Board) -> dict:
    return {"rules": f"rules for {board['brief']}"}

def coder(board: Board) -> dict:
    return {"code": f"game from {board['rules']}"}

builder = StateGraph(Board)
builder.add_node("designer", designer)
builder.add_node("coder", coder)
builder.add_edge(START, "designer")
builder.add_edge("designer", "coder")
builder.add_edge("coder", END)
graph = builder.compile()
print("result:", graph.invoke({"brief": "snake in space", "rules": "", "code": ""}))
```

## b · start-end-entry

`set_entry_point` is an older equivalent of an edge from `START` ; the docs use `add_edge`.

```python
entry = StateGraph(Board)
entry.add_node("designer", designer)
entry.set_entry_point("designer")
entry.set_finish_point("designer")
print("entry result:", entry.compile().invoke({"brief": "pong with planets", "rules": "", "code": ""}))
```
